```python
#!/usr/bin/env python3

import os
import sys
from pathlib import Path
import pandas as pd

NET_FILE = "D:\Thesis V2X\Project on Git\Berlin V2X Implementation\sumo1\osm1.net.xml"              # your SUMO network
CSV_FILE = "results\lat_lon.csv"              # your uploaded file
OUTPUT_ROUTE_FILE = "D:\Thesis V2X\Project on Git\Berlin V2X Implementation\sumo1\gps_route.rou.xml"

VCLASS = "passenger"
RADII = [50, 100, 150, 250]
MAX_CANDIDATES = 8
DOWNSAMPLE = 5

if "SUMO_HOME" not in os.environ:
    raise EnvironmentError("Please set SUMO_HOME")

sys.path.append(os.path.join(os.environ["SUMO_HOME"], "tools"))
import sumolib


def edge_allows(edge, vclass="passenger"):
    if edge.isSpecial():
        return False
    lanes = edge.getLanes()
    return any(l.allows(vclass) for l in lanes) if lanes else False


def get_candidate_edges(net, lat, lon, vclass="passenger"):
    x, y = net.convertLonLat2XY(lon, lat)

    candidates = []
    seen = set()

    for radius in RADII:
        nearby = net.getNeighboringEdges(x, y, r=radius, includeJunctions=False)
        nearby = [(e, d) for e, d in nearby if edge_allows(e, vclass)]
        nearby.sort(key=lambda t: t[1])  # SUMO does not sort automatically

        for edge, dist in nearby:
            eid = edge.getID()
            if eid not in seen:
                candidates.append((edge, dist))
                seen.add(eid)

        if len(candidates) >= MAX_CANDIDATES:
            break

    return candidates[:MAX_CANDIDATES]


def choose_edge(net, prev_edge, lat, lon, vclass="passenger"):
    candidates = get_candidate_edges(net, lat, lon, vclass=vclass)

    if not candidates:
        return None, None

    # first point: just take nearest valid edge
    if prev_edge is None:
        return candidates[0][0], None

    # later points: prefer a candidate reachable from previous edge
    best_edge = None
    best_path = None
    best_score = None

    for edge, snap_dist in candidates:
        if edge.getID() == prev_edge.getID():
            return edge, (prev_edge,)  # staying on same edge is ideal

        path, cost = net.getFastestPath(prev_edge, edge, vClass=vclass)

        if path is None:
            continue

        # combine routing cost and snap distance
        # lower is better
        score = cost + snap_dist

        if best_score is None or score < best_score:
            best_score = score
            best_edge = edge
            best_path = path

    return best_edge, best_path


def build_route(net, gps_points, vclass="passenger"):
    route = []
    prev_edge = None
    skipped = 0

    for i, (lat, lon) in enumerate(gps_points):
        edge, path = choose_edge(net, prev_edge, lat, lon, vclass=vclass)

        if edge is None:
            skipped += 1
            print(f"SKIP #{i}: no reachable candidate for ({lat}, {lon})")
            continue

        if prev_edge is None:
            route.append(edge)
            prev_edge = edge
            print(f"START #{i}: {edge.getID()}")
            continue

        if edge.getID() == prev_edge.getID():
            print(f"SAME  #{i}: {edge.getID()}")
            continue

        if path is None:
            # should only happen at start, but keep safe
            path = (edge,)

        print(f"LINK  #{i}: {prev_edge.getID()} -> {edge.getID()} via {len(path)} edges")

        for e in path[1:]:
            if route[-1].getID() != e.getID():
                route.append(e)

        prev_edge = edge

    print(f"\nFinal route edges: {len(route)}")
    print(f"Skipped GPS points: {skipped}")

    if len(route) < 2:
        raise RuntimeError("Could not build a connected route.")

    return route


def write_route(edges):
    edge_ids = " ".join(e.getID() for e in edges)
    xml = f"""<routes>
    <vType id="car" vClass="passenger"/>
    <route id="r0" edges="{edge_ids}"/>
    <vehicle id="veh0" type="car" route="r0" depart="0"/>
</routes>
"""
    Path(OUTPUT_ROUTE_FILE).write_text(xml, encoding="utf-8")
    print(f"Saved route to {OUTPUT_ROUTE_FILE}")


def main():
    print("Loading network...")
    net = sumolib.net.readNet(NET_FILE)

    print("Reading CSV...")
    df = pd.read_csv(CSV_FILE)
    gps_points = list(zip(df["Latitude"], df["Longitude"]))

    print(f"Loaded {len(gps_points)} GPS points")
    gps_points = gps_points[::DOWNSAMPLE]
    print(f"Using {len(gps_points)} points after downsampling by {DOWNSAMPLE}")

    print("Building continuity-aware route...")
    route_edges = build_route(net, gps_points, vclass=VCLASS)

    write_route(route_edges)


if __name__ == "__main__":
    main()
```
